## 0. Introducción y Contexto Geográfico, Histórico y Socioeconómico

En este informe analizamos la situación demográfica del estado Lara a partir de los datos del censo realizado en 2011 en Venezuela. El propósito del trabajo es describir de forma clara y directa cómo se distribuye la población en el territorio, su composición por edad y sexo, el nivel educativo de sus habitantes, la situación del mercado laboral y el comportamiento de los tres componentes vitales de la dinámica de la población: los nacimientos, las muertes y las migraciones.

### 0.1. Ubicación geográfica y organización del territorio

El estado Lara se encuentra localizado en la región centroccidental del país. Por su posición geográfica, ha funcionado históricamente como un punto de conexión natural y un lugar de paso entre el centro y el occidente venezolano, enlazando los Andes, los Llanos occidentales y la costa del Mar Caribe. Limita por el norte con el estado Falcón, por el sur con los estados Trujillo y Portuguesa, por el este con el estado Yaracuy y por el oeste con el estado Zulia. Cuenta con una superficie territorial de 19.800 kilómetros cuadrados, lo que representa el 2,16 % de la superficie total del país.

En el aspecto político y administrativo, el estado está dividido en nueve municipios autónomos:

* **Iribarren**, cuya capital Barquisimeto concentra la mayor parte de las actividades comerciales, administrativas, industriales y de servicios de la región.
* **Palavecino**, con su capital Cabudare, que ha crecido de manera acelerada en las últimas décadas como ciudad satélite muy articulada con Barquisimeto.
* **Torres**, con su capital Carora, el municipio más extenso del estado, abarcando más del 35 % del territorio larense.
* **Morán**, con capital en El Tocuyo, ciudad de gran trascendencia histórica y núcleo de la vida agrícola del valle del río Tocuyo.
* **Jiménez**, con capital en Quíbor, reconocido centro hortícola del país con una fuerte tradición artesanal.
* **Crespo**, con capital en Duaca, ubicado hacia el noreste en el piedemonte que conecta con las zonas agrícolas vecinas.
* **Andrés Eloy Blanco**, con capital en Sanare, enclavado en las estribaciones andinas con clima templado de montaña.
* **Urdaneta**, con capital en Siquisique, caracterizado por extensas zonas semiáridas y asentamientos rurales dispersos.
* **Simón Planas**, con capital en Sarare, que marca el paso geográfico y productivo hacia los Llanos occidentales.

### 0.2. Relieve, hidrografía y ejes socioeconómicos

El relieve larense presenta una amplia variedad de paisajes y relieves. Al sur y al sureste destacan las tierras altas pertenecientes a las estribaciones andinas y a la cordillera de la Costa, como las serranías de Sanare y Barbacoas, donde las alturas superan los dos mil metros sobre el nivel del mar. En contraste, en el centro y hacia el occidente predominan las depresiones de Carora y el valle de Quíbor, con valles aluviales y paisajes semiáridos con vegetación xerófila. Los principales cursos de agua son el río Tocuyo, que recorre la entidad de sur a norte antes de desembocar en el mar Caribe, y el río Turbio, que riega el valle donde se asientan Barquisimeto y Cabudare.

Esta diversidad geográfica ha condicionado una estructura productiva y social organizada alrededor de cuatro grandes ejes socioeconómicos:

1. **Eje Metropolitano de Servicios y Comercio (Barquisimeto - Cabudare):** Integrado por los municipios Iribarren y Palavecino, concentra el 65,94 % de la población del estado en apenas el 16,16 % del territorio físico. Barquisimeto funciona como el gran núcleo de servicios universitarios, centros hospitalarios especializados, banca y distribución comercial para todo el centroccidente del país, mientras que Cabudare alberga una densa área residencial cuyos habitantes se desplazan diariamente hacia la capital para trabajar o estudiar.
2. **Eje Agrícola de Riego y Valles Altos (Quíbor, El Tocuyo, Sanare y Duaca):** Agrupa a los municipios Jiménez, Morán, Andrés Eloy Blanco y Crespo. Quíbor sobresale como el principal centro hortícola del país con cultivos bajo riego de cebolla, tomate y pimentón, complementado por la producción artesanal de Tintorero. El Tocuyo sostiene una tradicional producción agrícola y de caña de azúcar con plantas de molienda, mientras que Sanare concentra la producción de café de sombra, papas y hortalizas de altura en parcelas familiares. Duaca aporta producción de hortalizas y frutales en la transición hacia el norte.
3. **Eje Ganadero y Semiárido Occidental (Carora y Siquisique):** Formado por los municipios Torres y Urdaneta. En Carora destaca la ganadería de doble propósito mediante el desarrollo de la raza bovina Carora, adaptada a las condiciones del trópico seco, junto con cultivos de caña de azúcar y viñedos en Altagracia. En Urdaneta, en cambio, la población se distribuye de forma dispersa en pequeños caseríos, dedicada principalmente a la cría de ganado caprino y al aprovechamiento de especies xerófilas en condiciones de marcada aridez.
4. **Eje Agroindustrial de Transición Llanera (Sarare):** Corresponde al municipio Simón Planas, caracterizado por sus tierras planas aptas para cultivos extensivos de maíz, sorgo y caña de azúcar, así como por la presencia de silos de almacenamiento y plantas procesadoras de alimentos para animales.

### 0.3. Aspectos metodológicos

Para realizar este estudio trabajamos con una base de datos relacional en SQLite (`demografia_lara.db`) estructurada a partir de los datos publicados y microdatos del censo de 2011 por el Instituto Nacional de Estadística. Se analiza la población residente habitual empadronada en la entidad, clasificada en grupos de edad de cinco en cinco años desde los 0 hasta los 95 años y más. A lo largo del informe examinamos cada cuadro y gráfico de forma directa y descriptiva, mostrando frecuencias, porcentajes, tasas y promedios, relacionando las cifras observadas con las características territoriales y socioeconómicas del estado Lara.


```python
import sqlite3
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['font.sans-serif'] = 'Helvetica'
plt.rcParams['axes.edgecolor'] = '#4A607A'
plt.rcParams['axes.linewidth'] = 0.8

COLOR_NAVY = '#1B365D'
COLOR_SLATE = '#4A607A'
COLOR_STEEL = '#7BA0C0'
COLOR_CRIMSON = '#A85A66'

# Permite ejecutar el cuaderno tanto desde notebooks/ como desde la raíz del proyecto
_db_candidates = [
    Path('data/processed/demografia_lara.db'),
    Path('../data/processed/demografia_lara.db'),
]
DB_PATH = next((p for p in _db_candidates if p.exists()), _db_candidates[0])
conn = sqlite3.connect(DB_PATH)

def fmt_ve(x, d=0):
    if pd.isna(x):
        return '-'
    if isinstance(x, (int, np.integer)):
        return f'{x:,.0f}'.replace(',', '.')
    if isinstance(x, (float, np.floating)):
        return f'{x:,.{d}f}'.replace(',', 'X').replace('.', ',').replace('X', '.')
    return str(x)

print(f'Base de datos conectada correctamente: {DB_PATH}')
```

## 1. Marco Político-Territorial, Crecimiento y Calidad Censal


```python
query_t1 = '''
SELECT ubigeo, municipio, capital, subregion_funcional, superficie_km2, 
       pct_superficie_estadal, poblacion_2011, pct_poblacion_2011, 
       densidad_2011_hab_km2, pct_urbana_2011 
FROM territorio_superficie_densidad_lara 
ORDER BY CAST(ubigeo AS INTEGER)
'''
df_t1 = pd.read_sql_query(query_t1, conn)

df_t1_disp = df_t1.copy()
df_t1_disp['superficie_km2'] = df_t1_disp['superficie_km2'].apply(lambda x: fmt_ve(x, 1))
df_t1_disp['pct_superficie_estadal'] = df_t1_disp['pct_superficie_estadal'].apply(lambda x: fmt_ve(x, 2) + '%')
df_t1_disp['poblacion_2011'] = df_t1_disp['poblacion_2011'].apply(lambda x: fmt_ve(x, 0))
df_t1_disp['pct_poblacion_2011'] = df_t1_disp['pct_poblacion_2011'].apply(lambda x: fmt_ve(x, 2) + '%')
df_t1_disp['densidad_2011_hab_km2'] = df_t1_disp['densidad_2011_hab_km2'].apply(lambda x: fmt_ve(x, 2))
df_t1_disp['pct_urbana_2011'] = df_t1_disp['pct_urbana_2011'].apply(lambda x: fmt_ve(x, 1) + '%')

df_t1_disp.columns = ['Código', 'Municipio', 'Capital', 'Subregión', 'Superficie (km²)', '% Superficie', 'Población 2011', '% Población', 'Densidad (hab/km²)', '% Urbano']
print('Cuadro 1.1: Organización Territorial, Superficie y Densidad Demográfica del Estado Lara (2011)')
display(df_t1_disp)
print('Fuente: Instituto Nacional de Estadística (INE), censo de 2011. Cálculos propios.')
```

La ocupación del territorio larense responde a una asimetría estructural persistente. La distribución municipal del empadronamiento (Cuadro 1.1) expone una fractura espacial severa: el eje metropolitano conformado por Iribarren (56,13 % de los habitantes, con 996.230 personas) y Palavecino (9,81 %, con 174.099 personas) concentra el 65,94 % de la población total sobre apenas el 16,16 % de la superficie estatal, registrando densidades críticas de 360,95 hab/km² y 395,68 hab/km² respectivamente. El espacio restante acusa un vaciamiento demográfico sistemático. Torres y Urdaneta abarcan conjuntamente el 55,71 % de la geografía física de Lara —con 6.937,0 km² y 4.093,0 km²—, pero retienen apenas el 13,90 % de los residentes (185.275 y 61.381 habitantes) bajo densidades de 26,71 hab/km² y 15,00 hab/km². Esta disparidad obedece al confinamiento agroclimático del trópico seco: la aridez en la depresión de Carora y el valle medio del río Tocuyo impone restricciones ecológicas persistentes a la ocupación humana dispersa, empujando la fuerza de trabajo hacia la red de hospitales Tipo IV, universidades y nodos de acopio comercial de Barquisimeto-Cabudare. El espacio agrícola intermedio sostiene densidades moderadas: los valles de Morán (6,98 %), Jiménez (5,69 %), Crespo (2,81 %) y Andrés Eloy Blanco (2,66 %) articulan sistemas de riego y caficultura de altura, mientras que Simón Planas concentra el 2,02 % en la agroindustria de Sarare.


```python
query_t12 = '''
SELECT periodo, tiempo_t, poblacion_inicial_entidad, poblacion_final_entidad, 
       crecimiento_absoluto_entidad, tasa_geometrica_entidad, 
       tasa_exponencial_entidad, poblacion_inicial_vzla, 
       poblacion_final_vzla, tasa_geometrica_vzla, 
       tasa_exponencial_vzla 
FROM censo_historico_intercensal
'''
df_t12 = pd.read_sql_query(query_t12, conn)

df_t12_disp = df_t12.copy()
df_t12_disp['poblacion_inicial_entidad'] = df_t12_disp['poblacion_inicial_entidad'].apply(lambda x: fmt_ve(x, 0))
df_t12_disp['poblacion_final_entidad'] = df_t12_disp['poblacion_final_entidad'].apply(lambda x: fmt_ve(x, 0))
df_t12_disp['crecimiento_absoluto_entidad'] = df_t12_disp['crecimiento_absoluto_entidad'].apply(lambda x: fmt_ve(x, 0))
df_t12_disp['tasa_geometrica_entidad'] = df_t12_disp['tasa_geometrica_entidad'].apply(lambda x: fmt_ve(x, 2) + '%')
df_t12_disp['tasa_exponencial_entidad'] = df_t12_disp['tasa_exponencial_entidad'].apply(lambda x: fmt_ve(x, 2) + '%')
df_t12_disp['poblacion_inicial_vzla'] = df_t12_disp['poblacion_inicial_vzla'].apply(lambda x: fmt_ve(x, 0))
df_t12_disp['poblacion_final_vzla'] = df_t12_disp['poblacion_final_vzla'].apply(lambda x: fmt_ve(x, 0))
df_t12_disp['tasa_geometrica_vzla'] = df_t12_disp['tasa_geometrica_vzla'].apply(lambda x: fmt_ve(x, 2) + '%')
df_t12_disp['tasa_exponencial_vzla'] = df_t12_disp['tasa_exponencial_vzla'].apply(lambda x: fmt_ve(x, 2) + '%')

df_t12_disp.columns = ['Período', 'Años (t)', 'Pob. Inicial Lara', 'Pob. Final Lara', 'Crec. Abs. Lara', 'r geom Lara (%)', 'r exp Lara (%)', 'Pob. Inicial Vzla', 'Pob. Final Vzla', 'r geom Vzla (%)', 'r exp Vzla (%)']
print('Cuadro 1.2: Crecimiento Intercensal de la Población del Estado Lara y Venezuela (1981–2011)')
display(df_t12_disp)
print('Fuente: Instituto Nacional de Estadística (INE), censos de población de 1981, 1990, 2001 y 2011.')
```

```python
anios_censo = [1981, 1990, 2001, 2011]
pob_lara = [944475, 1193161, 1556415, 1774867]

fig, ax = plt.subplots(figsize=(8, 4.8), dpi=120)

bars = ax.bar(anios_censo, [p/1000.0 for p in pob_lara], width=3.2, color=COLOR_STEEL, alpha=0.55, label='Población Censada (Miles de Habitantes)')
ax.plot(anios_censo, [p/1000.0 for p in pob_lara], color=COLOR_NAVY, marker='o', linewidth=2.2, markersize=7, label='Tendencia Intercensal')

for x, p in zip(anios_censo, pob_lara):
    ax.text(x, (p/1000.0) + 40, f'{fmt_ve(p, 0)} hab.', ha='center', va='bottom', fontsize=9, fontweight='bold', color=COLOR_NAVY)

ax.set_ylim(700, 2100)
ax.set_xticks(anios_censo)
ax.set_xticklabels([str(a) for a in anios_censo], fontsize=10)
ax.set_xlabel('Año Censal', fontsize=10, color=COLOR_SLATE)
ax.set_ylabel('Población Censada (Miles de Habitantes)', fontsize=10, color=COLOR_SLATE)
ax.set_title('Gráfico 1.1. Estado Lara: Evolución de la Población Total Censada, 1981–2011', fontsize=11, fontweight='bold', color=COLOR_NAVY, pad=12)
ax.legend(loc='upper left', frameon=False)
ax.grid(axis='y', linestyle='--', alpha=0.3)

fig.text(0.12, -0.02, 'Fuente: Instituto Nacional de Estadística (INE), censos de población (1981--2011). Base de datos demografia_lara.db.', fontsize=8, color='#555555')

plt.tight_layout()
plt.show()
```

El dinamismo intercensal larense perdió velocidad de forma contundente tras décadas de fuerte expansión. La trayectoria demográfica entre 1981 y 2011 (Cuadro 1.2 y Gráfico 1.1) registra un incremento absoluto de 944.475 a 1.774.867 habitantes —un saldo neto acumulado de 830.392 personas en treinta años—, pero esconde un frenazo abrupto en el ritmo de crecimiento de la entidad. La tasa de expansión anual se contrajo a la mitad. Las tasas exponenciales intercensales de Lara pasaron de 2,55 % anual entre 1981 y 1990 (frente a 2,41 % en Venezuela) y 2,42 % entre 1990 y 2001 (frente a 2,20 % nacional) a un 1,31 % en el decenio 2001-2011, quedando por debajo del 1,66 % registrado por la República. La transición demográfica pisó el freno. La reducción simultánea de los niveles de natalidad y el debilitamiento de los saldos migratorios positivos que alimentaron la expansión metropolitana barquisimetana en el auge petrolero explican el agotamiento del ritmo expansivo previo.


## 2. Estructura y Composición por Edad y Sexo


```python
query_p = '''
SELECT orden_edad, grupo_edad, hombres, mujeres, total, 
       pct_hombres, pct_mujeres, pct_total, indice_masculinidad 
FROM piramide_quinquenal_edad_sexo_lara 
WHERE codigo_ubigeo='13' AND anio=2011 
ORDER BY orden_edad
'''
df_pir = pd.read_sql_query(query_p, conn)

df_pir_disp = df_pir.copy()
df_pir_disp['hombres'] = df_pir_disp['hombres'].apply(lambda x: fmt_ve(x, 0))
df_pir_disp['mujeres'] = df_pir_disp['mujeres'].apply(lambda x: fmt_ve(x, 0))
df_pir_disp['total'] = df_pir_disp['total'].apply(lambda x: fmt_ve(x, 0))
df_pir_disp['pct_hombres'] = df_pir_disp['pct_hombres'].apply(lambda x: fmt_ve(x, 2) + '%')
df_pir_disp['pct_mujeres'] = df_pir_disp['pct_mujeres'].apply(lambda x: fmt_ve(x, 2) + '%')
df_pir_disp['pct_total'] = df_pir_disp['pct_total'].apply(lambda x: fmt_ve(x, 2) + '%')
df_pir_disp['indice_masculinidad'] = df_pir_disp['indice_masculinidad'].apply(lambda x: fmt_ve(x, 2))

df_pir_disp.columns = ['Orden', 'Grupo de Edad', 'Hombres', 'Mujeres', 'Total', '% Hombres', '% Mujeres', '% Grupo', 'Índice Masculinidad']
print('Cuadro 2.1: Población por Grupos Quinquenales de Edad y Razón de Masculinidad (Estado Lara, 2011)')
display(df_pir_disp[['Grupo de Edad', 'Hombres', 'Mujeres', 'Total', '% Hombres', '% Mujeres', '% Grupo', 'Índice Masculinidad']])
print('Fuente: Instituto Nacional de Estadística (INE), censo de 2011. Cálculos propios.')
```

```python
fig, ax = plt.subplots(figsize=(8, 6.2), dpi=120)

y_pos = np.arange(len(df_pir))
hombres_val = df_pir['hombres'].values / 1000.0
mujeres_val = df_pir['mujeres'].values / 1000.0

ax.barh(y_pos, -hombres_val, height=0.78, color=COLOR_NAVY, alpha=0.85, label='Hombres')
ax.barh(y_pos, mujeres_val, height=0.78, color=COLOR_STEEL, alpha=0.85, label='Mujeres')

ax.set_yticks(y_pos)
ax.set_yticklabels(df_pir['grupo_edad'], fontsize=8.5)
ax.set_xlabel('Población Empadronada (Miles de Habitantes)', fontsize=10, color=COLOR_SLATE)
ax.set_ylabel('Grupos Quinquenales de Edad', fontsize=10, color=COLOR_SLATE)
ax.set_title('Gráfico 2.1. Estado Lara: Pirámide de Población por Grupos Quinquenales (Censo 2011)', fontsize=11, fontweight='bold', color=COLOR_NAVY, pad=12)

ticks = np.linspace(-100, 100, 9)
ax.set_xticks(ticks)
ax.set_xticklabels([f'{abs(int(t))}' for t in ticks], fontsize=9)

ax.axvline(0, color='#4A607A', linewidth=0.8)
ax.legend(loc='upper right', frameon=False)
ax.grid(axis='x', linestyle='--', alpha=0.3)

fig.text(0.12, -0.02, 'Fuente: Instituto Nacional de Estadística (INE), censo de 2011. Base de datos demografia_lara.db.', fontsize=8, color='#555555')

plt.tight_layout()
plt.show()
```

La base piramidal ya se estancó. La distribución por grupos quinquenales (Cuadro 2.1 y Gráfico 2.1) constata el estrechamiento de los primeros escalones etarios: el tramo de 0 a 4 años registra 155.691 niños (8,77 %) y el de 5 a 9 años contabiliza 155.634 personas (8,77 %), montos inferiores a los picos de 10 a 14 años (166.892 personas, 9,40 %) y 15 a 19 años (170.435 personas, 9,60 %). El ensanchamiento central domina la estructura. El volumen en edades activas (15 a 64 años) alcanza 1.187.304 personas —el 66,90 % del total estadal—, configurando una razón de dependencia total de 49,49 dependientes por cada 100 activos, descompuesta en una razón de dependencia infantil de 40,28 % y una razón de dependencia de vejez de 9,21 %, con un índice de envejecimiento situado en 22,87 % y una edad mediana de 27,54 años. El sesgo de masculinidad cambia radicalmente a lo largo del ciclo vital. Frente al índice de masculinidad global de 99,18 varones por cada 100 mujeres —correspondiente a 883.775 hombres y 891.092 mujeres empadronadas—, la ventaja biológica natalicia en 0-4 años (106,31) y 5-9 años (106,41) se diluye paulatinamente en las edades adultas y se derrumba en la ancianidad, cayendo a 71,27 en 80-84 años y a 55,05 en mayores de 95 años como consecuencia de la sobremortalidad diferencial acumulada.

## 6. Conclusiones Generales

La configuración territorial del estado Lara reproduce una polarización macrocefálica persistente entre el núcleo metropolitano y su periferia semiárida. La conurbación funcional integrada por Iribarren y Palavecino reúne al 65,94 % de la población estatal (1.170.329 habitantes) en un estrecho 16,16 % del territorio físico (3.200,0 km²), imponiendo densidades críticas que oscilan entre 360,95 hab/km² y 395,68 hab/km² gracias al monopolio de la red de hospitales Tipo IV, los recintos universitarios y el aparato de comercialización mayorista centroccidental. Esta hiperconcentración contrasta con el vaciamiento relativo de las extensiones xerófilas de Torres (26,71 hab/km²) y Urdaneta (15,00 hab/km²), donde el 55,71 % de la geografía larense retiene apenas al 13,90 % de los residentes bajo severas restricciones hídricas que limitan el poblamiento permanente. El drenaje histórico hacia el exterior pisó el freno. La pérdida acumulada de toda la vida de 45.652 personas (-1,82 ‰ anual) dio paso en el quinquenio 2006-2011 a un saldo neto negativo de apenas 5.206 individuos (-0,61 ‰ anual), confirmando que Barquisimeto y Cabudare consolidaron un polo de atracción regional capaz de retener fuerza de trabajo y neutralizar la expulsión neta hacia Caracas o la cuenca petrolera del Zulia.

La pirámide demográfica larense exhibe un estancamiento definitivo en su base generacional que confirma el avance de una transición etaria madura. Los tramos de 0 a 4 años (155.691 niños, 8,77 %) y de 5 a 9 años (155.634 infantes, 8,77 %) se sitúan por debajo de las cohortes de 10 a 14 años (9,40 %) y 15 a 19 años (9,60 %), como reflejo del descenso de la Tasa Global de Fecundidad a 2,06 hijos por mujer y una Tasa Bruta de Reproducción de 1,01 hijas que roza el límite de reemplazo biológico. Este estrechamiento basal traslada el volumen hacia las edades activas (15 a 64 años), que agrupan al 66,90 % de los habitantes y reducen la razón de dependencia demográfica total a 49,49 %, consolidando una ventana óptima de bono demográfico. La presión productiva inmediata resulta mayúscula. Las proyecciones hacia 2016 revelan que la incorporación neta de 66.315 trabajadores a la PEA exigirá generar 61.010 empleos ocupados para mantener la tasa de desocupación en el 8,00 % estructural, enfrentando el desafío de que el 42,00 % de los ocupados (332.753 personas) continuará absorbido por la precariedad del sector informal si no median transformaciones en la estructura productiva regional.

El balance de la mortalidad larense encierra una paradoja epidemiológica que solo la estandarización demográfica logra dilucidar. La tasa bruta de mortalidad observada de 5,14 ‰ supera la media nacional de 5,02 ‰, pero este diferencial obedece exclusivamente a una estructura poblacional con mayor proporción de adultos mayores; al tipificar directamente contra la población estándar de Venezuela, la tasa ajustada larense desciende a 4,96 ‰, ratificando condiciones de supervivencia intrínseca más favorables que el promedio de la República. Más de la mitad de los decesos del estado provienen de afecciones crónicas asociadas al envejecimiento —con las enfermedades del corazón, el cáncer, los accidentes cerebrovasculares y la diabetes sumando el 51,90% del total—, pero la estructura de riesgos se distorsiona en los adultos jóvenes por la incidencia de colisiones terrestres y homicidios (10,30% conjunto), cuya concentración en los varones explica la brecha de 6,08 años entre la esperanza de vida femenina (77,56 años) y la masculina (71,48 años).
